## 1.2 Distribution de la cible : `is_substitutable`

**Pourquoi ce graphique :**  
Visualiser le déséquilibre entre les classes est **la première analyse indispensable** avant d'entraîner un classifieur. Un déséquilibre non pris en compte biaise le modèle vers la classe majoritaire.

**Ce qu'on cherche ici :**
- Mesurer le ratio substituable / non-substituable
- Décider si on a besoin de rééquilibrage (SMOTE, pondération, stratification)

**Décision prise :** ~84% substituables / 16% non-substituables → déséquilibre modéré (label v2 sigmoid).  
On applique `class_weight='balanced'` sur tous les classifieurs et on stratifie le split train/test pour conserver les proportions.

**Note sur l'évolution du label (v1 → v2) :**  
- **v1** (obsolète) : `is_substitutable = 1` si `distance_km ≤ 600` ET vol existant. Résultat : 89.5% substituables — trop mécanique, le modèle mémorisait la règle (F1=1.000 trivial).
- **v2** (actuel) : scoring sigmoid pondéré par service ferroviaire avec 600km comme **pivot** (pas seuil dur). Résultat : 84.3% substituables — 5 190 corridors < 600km sans service → non-substituables, 2 806 corridors > 600km bien desservis → substituables.

**Choix de la métrique principale :**  
Dans le contexte d'ObRail, les deux types d'erreurs n'ont pas le même coût métier :
- **Faux positif (FP)** : recommander un corridor "non-substituable" comme substituable → impact limité
- **Faux négatif (FN)** : rater un corridor réellement substituable → **coût environnemental réel**

**→ On surveille en priorité le Recall de la classe 1 et le F1 weighted.**

````python
counts = df['is_substitutable'].value_counts()
labels = ['Non substituable (0)', 'Substituable (1)']
colors = ['#e74c3c', '#2ecc71']

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(labels, [counts[0], counts[1]], color=colors, edgecolor='white')
axes[0].set_title('Distribution is_substitutable')
axes[0].set_ylabel('Nombre de corridors')
for i, v in enumerate([counts[0], counts[1]]):
    axes[0].text(i, v + 5, str(v), ha='center', fontweight='bold')

axes[1].pie([counts[0], counts[1]], labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
axes[1].set_title('Répartition (%)')

plt.suptitle(f'Déséquilibre : {counts[1]/len(df)*100:.1f}% substituables / {counts[0]/len(df)*100:.1f}% non-substituables')
plt.tight_layout()
plt.savefig('../docs/fig_is_substitutable.png', dpi=150, bbox_inches='tight')
plt.show()
print('Déséquilibre → utiliser class_weight="balanced" dans les modèles')
````

**Ce qu'on cherche ici :**

## 1.6 Matrice de corrélations

**Pourquoi cette heatmap :**  
La matrice de corrélations sert à deux choses essentielles avant la modélisation :

1. **Détecter la multicolinéarité** : si deux features sont très corrélées entre elles (|r| > 0.9), en garder une seule suffit (les arbres gèrent mieux, mais la Ridge Regression peut souffrir)
2. **Lire le positionnement de chaque variable vis-à-vis de la cible** `is_substitutable`

**Ce qu'on observe — les deux « bras » du label v2 :**

Le label v2 compare un `service_score` à un seuil sigmoïde piloté par la distance. La matrice rend visibles ces deux mécanismes opposés :

- **Bras « desserte » (positif)** : `service_share` **+0.28** (2ᵉ corrélation la plus forte de tout le tableau), `trip_count_corridor` **+0.16**, `ratio_origin`/`ratio_dest` ≈ +0.05. Plus un corridor est desservi, plus il est substituable → c'est le **numérateur** du score.
- **Bras « distance / CO2 » (négatif)** : `co2_avion_kg` **−0.33**, `distance_km` **−0.31**, `co2_saved_kg` **−0.31**, `co2_train_kg` −0.20. Plus c'est long, plus le seuil sigmoïde monte, moins c'est substituable → c'est le **seuil** à franchir.
- **Trafic et population bruts ≈ 0** (`origin_station_traffic` +0.02, `origin_city_population` 0.00) : aucun signal linéaire direct. C'est exactement pour ça qu'on les a transformés en `ratio` et en `service_share` — ce sont les features **dérivées**, pas les valeurs brutes, qui portent l'information.

**Multicolinéarité :** `co2_saved_kg ↔ co2_avion_kg` = 0.98 et `distance_km ↔ co2_avion_kg` = 0.90 → bloc redondant (même physique sous-jacente). `origin_station_traffic ↔ origin_city_population` = 0.69 (les gares des grandes villes sont plus fréquentées). En revanche `service_share` et `trip_count_corridor` ne sont **pas** colinéaires avec la distance → elles apportent une information *indépendante*, ce qu'on recherche pour le M1.

**Nuance à garder en tête :** toutes ces corrélations restent modérées (|r| ≤ 0.33) car le label v2 est **non-linéaire** (termes saturants, ratio, seuil sigmoïde) et le coefficient de Pearson sous-estime ce type de relation. Le classement reste néanmoins cohérent : `service_share` 2ᵉ ici annonce la hiérarchie qu'on retrouvera, amplifiée, dans les **importances du Modèle 1** (`trip_count_corridor` 47.9 %, `service_share` 11.6 %).

**Décision prise :** Le bloc CO2/distance fortement corrélé confirme le **risque de biais circulaire dans M2** si on incluait `co2_avion_kg`/`co2_train_kg`. On les exclut donc des features du Modèle 2 — le modèle doit apprendre depuis la géographie et la desserte, pas depuis les valeurs CO2 calculées.

````python
corr_cols = ['distance_km','co2_train_kg','co2_avion_kg','co2_saved_kg',
             'trip_count_corridor','service_share','ratio_origin','ratio_dest',
             'origin_station_traffic','origin_city_population','is_substitutable']
corr = df[corr_cols].corr()

fig, ax = plt.subplots(figsize=(11, 9))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn', center=0,
            mask=mask, ax=ax, linewidths=0.5, vmin=-1, vmax=1)
ax.set_title('Matrice de corrélations')
plt.tight_layout()
plt.savefig('../docs/fig_correlations.png', dpi=150, bbox_inches='tight')
plt.show()

print('Corrélations avec is_substitutable :')
print(corr['is_substitutable'].drop('is_substitutable').sort_values().round(3))
````

## 1.7 Valeurs manquantes

**Pourquoi cette analyse :**  
Chaque valeur manquante est une décision de modélisation. Il n'existe pas de stratégie universelle : on choisit en fonction du **sens métier** de chaque colonne.

**Stratégies retenues :**

| Colonne | Stratégie | Justification |
|---|---|---|
| `co2_avion_kg` / `co2_saved_kg` | **Exclure du Modèle 2** | NULL = pas de vol → l'exemple est invalide pour la régression |
| `*_station_traffic` | **Remplacer par 0** | NULL = gare non-française, pas de données SNCF disponibles |
| `*_city_population` | **Remplacer par 0** | NULL = ville non reconnue, pas un zéro réel |
| `service_share` / `trip_count_*` | **Remplacer par 0** | NULL = gare non couverte par GTFS |
| `distance_km` / `co2_train_kg` | **Exclure** | NULL = corridor invalide, données ETL incomplètes |

**Choix de remplacement par 0 vs médiane :**  
On choisit 0 et non la médiane car 0 a un **sens métier clair** : "nous n'avons pas d'information sur cette gare". Les modèles à base d'arbres (Random Forest, XGBoost) apprennent à interpréter ce 0 comme une absence d'information, ce qu'ils font naturellement via leurs seuils de décision.

# SECTION 2 — Préparation des données ML

## 2.1 Sélection et nettoyage des features

**Pourquoi 14 features pour M1, 11 pour M2 :**

**Modèle 1 (Classification — is_substitutable) :**  
On inclut `co2_avion_kg` et `co2_train_kg` car :
- `co2_avion_kg != 0` est le proxy de "un vol existe sur ce corridor" → information légitime
- Le modèle généralise en Europe où la règle des 600km n'est pas une loi — il combine distance + vol + démographie

Feature ajoutée v2 : **`dist_to_600 = distance_km − 600`** — encode explicitement la position par rapport au pivot légal français. Le modèle peut ainsi apprendre des règles asymétriques de part et d'autre du seuil légal.

**Modèle 2 (Régression — co2_saved_kg) :**  
On **exclut** `co2_avion_kg`, `co2_train_kg` et `dist_to_600` car :
- `co2_saved_kg = co2_avion_kg - co2_train_kg` → biais circulaire trivial si on les inclut (R²=1 garanti)
- `dist_to_600` est un encodage du pivot légal FR, sans lien direct avec le gain CO2 physique
- Le modèle doit apprendre depuis la **géographie et la démographie**, pas depuis les valeurs CO2 calculées

**Label v2 — scoring sigmoid (pivot 600km) :**  
```
service_score = 0.40 × min(service_share/0.12, 1)
              + 0.30 × min(trip_count_corridor/30, 1)
              + 0.30 × min((ratio_origin+ratio_dest)/20, 1)

seuil = 0.05 + 0.85 / (1 + exp(−0.007 × (distance_km − 600)))

is_substitutable = 1  si service_score ≥ seuil ET co2_avion_kg > 0
                   0  si distance_km > 1100km (limite physique absolue)
```
→ Le 600km est le point d'inflexion du seuil, pas un critère binaire. En dessous : seuil faible (~0.05–0.30). Au-dessus : seuil élevé (~0.50–0.90).

**Stratégie NULL → 0 :**  
Choisie pour les features de fréquentation, population et service car les modèles à base d'arbres (Random Forest, XGBoost) apprennent naturellement que "0 = donnée absente" via leurs seuils de décision.

````python
# ratio_origin / ratio_dest : trafic gare / population ville — features calculées, non présentes dans le CSV
df['ratio_origin'] = (df['origin_station_traffic'] / df['origin_city_population'].replace(0, float('nan'))).fillna(0)
df['ratio_dest']   = (df['dest_station_traffic']   / df['dest_city_population'].replace(0, float('nan'))).fillna(0)

# dist_to_600 : écart au pivot légal français (600km) — encode la position par rapport à la loi FR 2023
df['dist_to_600'] = df['distance_km'] - 600

# Features retenues pour les deux modèles
FEATURES_M1 = [
    'distance_km',
    'dist_to_600',                # pivot légal français explicite (v2)
    'co2_train_kg',
    'co2_avion_kg',
    'vehicule_type',              # → encodage LabelEncoder
    'origin_station_traffic',
    'origin_city_population',
    'dest_station_traffic',
    'dest_city_population',
    'ratio_origin',               # trafic/population départ
    'ratio_dest',                 # trafic/population arrivée
    'trip_count_corridor',        # trajets hebdo sur le corridor — feature #1 XGBoost v2
    'trip_count_origin',          # trajets hebdo total gare départ
    'service_share',              # trip_count_corridor / trip_count_origin
]
# M2 exclut co2_avion_kg, co2_train_kg (biais circulaire) et dist_to_600 (pivot légal FR sans lien CO2 physique)
FEATURES_M2 = [f for f in FEATURES_M1 if f not in ['co2_avion_kg', 'co2_train_kg', 'dist_to_600']]

TARGET_M1 = 'is_substitutable'
TARGET_M2 = 'co2_saved_kg'

df_ml = df[FEATURES_M1 + [TARGET_M1, TARGET_M2]].copy()

# Remplacement des NULL par 0 pour trafic, population, ratios et fréquences
fill_zero_cols = ['origin_station_traffic','origin_city_population',
                  'dest_station_traffic','dest_city_population',
                  'ratio_origin','ratio_dest',
                  'trip_count_corridor','trip_count_origin','service_share']
for col in fill_zero_cols:
    if col in df_ml.columns:
        df_ml[col] = df_ml[col].fillna(0)

# co2_avion_kg : NULL = pas de vol → remplacer par 0 pour le Modèle 1
df_ml['co2_avion_kg'] = df_ml['co2_avion_kg'].fillna(0)

print(f'Dataset ML : {len(df_ml)} lignes, {len(df_ml.columns)} colonnes')
print(f'Features M1 : {len(FEATURES_M1)} | Features M2 : {len(FEATURES_M2)}')
print(f'Valeurs manquantes restantes : {df_ml.isnull().sum().sum()}')
````

## 2.2 Encodage de `vehicule_type`

**Pourquoi LabelEncoder et non OneHotEncoder :**

`vehicule_type` a 15 modalités : AVE, AVE Nuit, EuroCity, EuroCity Nuit, EuroNight, ICE, ICE Nuit, InterCity, InterCity Nuit, Nightjet, TGV, TGV Nuit, Train Longue Distance, Train Longue Distance Nuit, Train Nuit.

**OneHotEncoder** créerait 15 nouvelles colonnes binaires — adapté pour la régression logistique ou les réseaux de neurones où l'ordre n'a pas de sens.

**LabelEncoder** attribue un entier (0→4) — adapté pour les **arbres de décision** car :
- Un arbre peut apprendre des seuils comme "type < 3" → sépare les trains de nuit des trains de jour
- Il n'introduit pas de présupposé d'ordre entre les catégories (les arbres utilisent des seuils, pas des distances)
- Réduit la dimensionnalité (1 colonne au lieu de 15)

**Important :** Le LabelEncoder est **sauvegardé** (`models/label_encoder_vehicule.joblib`) pour être réutilisé à l'identique lors des prédictions API. Sans ça, le mapping 0↔EuroNight peut changer à chaque réentraînement.

````python
le = LabelEncoder()
df_ml['vehicule_type_enc'] = le.fit_transform(df_ml['vehicule_type'])

print('Mapping vehicule_type :')
for i, cls in enumerate(le.classes_):
    print(f'  {i} → {cls}')

# On remplace la colonne texte par l'encodage
df_ml = df_ml.drop(columns=['vehicule_type'])
df_ml = df_ml.rename(columns={'vehicule_type_enc': 'vehicule_type'})
````

## 2.3 Dataset Modèle 1 — Classification `is_substitutable`

**Pourquoi stratifier le split :**  
Avec ~85% / 15% de déséquilibre, un split aléatoire non-stratifié pourrait créer un test set avec 100% de classe 1 par chance. La stratification garantit que **chaque split (train / val / test) préserve le ratio 85/15** — condition nécessaire pour que les métriques de test soient représentatives.

**Pourquoi 70/15/15 et non 80/20 :**  
Avec 46 106 corridors, le découpage 70/15/15 donne 32 274 / 6 916 / 6 916 exemples. Le test set (~6 916 corridors) donne des intervalles de confiance très resserrés sur le F1, et le set de validation (~6 916) est largement suffisant pour l'early stopping et le choix des hyperparamètres.

**Seed `random_state=42` :**  
Fixé partout pour garantir la **reproductibilité** : deux runs successifs produisent exactement les mêmes splits, ce qui permet de comparer les modèles sur strictement les mêmes données.

````python
X_m1 = df_ml[FEATURES_M1]
y_m1 = df_ml[TARGET_M1]

X_train_m1, X_temp, y_train_m1, y_temp = train_test_split(
    X_m1, y_m1, test_size=0.30, random_state=42, stratify=y_m1
)
X_val_m1, X_test_m1, y_val_m1, y_test_m1 = train_test_split(
    X_temp, y_temp, test_size=0.50, random_state=42, stratify=y_temp
)

print('=== Modèle 1 — Classification ===')
print(f'Train : {len(X_train_m1)} lignes | Val : {len(X_val_m1)} | Test : {len(X_test_m1)}')
print(f'Train is_sub=1 : {y_train_m1.mean()*100:.1f}% | Test is_sub=1 : {y_test_m1.mean()*100:.1f}%')
````

## 2.4 Dataset Modèle 2 — Régression `co2_saved_kg`

**Pourquoi filtrer sur les corridors avec vol :**  
`co2_saved_kg = co2_avion_kg - co2_train_kg` n'est défini que si un vol existe (`co2_avion_kg` non NULL). Pour les corridors sans vol, la question "combien de CO2 économisé ?" n'a pas de sens — il n'y a pas d'avion à remplacer.

**Conséquence :** Le Modèle 2 est entraîné sur **43 782 corridors** (le sous-ensemble des 46 106 ayant un vol direct, split 30 647 / 6 567 / 6 568) — uniquement ceux qui ont un vol existant.

**Lien M1 → M2 dans le pipeline de prédiction :**  
En production, on exécute toujours M1 d'abord. Si `is_substitutable = 0`, on n'exécute pas M2 (inutile). Si `is_substitutable = 1`, on exécute M2 pour quantifier le gain. Cela correspond à la logique de l'endpoint `/predict/co2_saved` de l'API.

````python
# Pour la régression : on garde uniquement les corridors avec un vol existant
df_m2 = df_ml[df[TARGET_M2].notna()].copy()
df_m2[TARGET_M2] = df[TARGET_M2][df[TARGET_M2].notna()].values

X_m2 = df_m2[FEATURES_M1]
y_m2 = df_m2[TARGET_M2]

X_train_m2, X_temp, y_train_m2, y_temp = train_test_split(
    X_m2, y_m2, test_size=0.30, random_state=42
)
X_val_m2, X_test_m2, y_val_m2, y_test_m2 = train_test_split(
    X_temp, y_temp, test_size=0.50, random_state=42
)

print('=== Modèle 2 — Régression ===')
print(f'Lignes utilisables (avec vol) : {len(df_m2)}')
print(f'Train : {len(X_train_m2)} | Val : {len(X_val_m2)} | Test : {len(X_test_m2)}')
print(f'co2_saved_kg — moy: {y_m2.mean():.1f} | méd: {y_m2.median():.1f} | std: {y_m2.std():.1f}')
````

## 2.5 Normalisation des features numériques

**Pourquoi StandardScaler :**  
Les features ont des échelles très différentes :
- `distance_km` : [100 — 918]
- `origin_station_traffic` : [0 — 257 000 000]
- `service_share` : [0.0 — 1.0]

Sans normalisation, les modèles sensibles aux distances (Logistic Regression, MLP, Ridge) seraient dominés par les grandes valeurs. `station_traffic` à 258M écraserait `service_share` à 0.05.

**StandardScaler vs MinMaxScaler :**  
- `StandardScaler` : moyenne=0, écart-type=1. **Robuste aux valeurs aberrantes** dans la plage, préserve la forme de la distribution
- `MinMaxScaler` : compresse entre [0,1]. Très sensible aux outliers (un trafic de 258M tire tout vers le bas)

On choisit StandardScaler car nos features de trafic ont des distributions à longue queue (quelques grandes gares).

**Pourquoi les arbres n'ont pas besoin de normalisation :**  
Random Forest et XGBoost utilisent des **seuils de comparaison** (`distance_km < 350?`) et non des distances euclidiennes. La normalisation n'affecte pas leurs performances — mais on la conserve pour cohérence avec Logistic Regression et MLP dans la comparaison.

**Règle d'or :** On `fit` le scaler **uniquement sur le train**, puis `transform` sur val et test. Sinon, on introduit de la **data leakage** (le modèle "voit" la distribution du test avant l'évaluation).

````python
# StandardScaler : moyenne=0, écart-type=1
# On fit sur le train uniquement, on applique sur val et test
scaler = StandardScaler()

X_train_m1_scaled = scaler.fit_transform(X_train_m1)
X_val_m1_scaled   = scaler.transform(X_val_m1)
X_test_m1_scaled  = scaler.transform(X_test_m1)

print('Normalisation appliquée (StandardScaler)')
print(f'Moyenne train avant : {X_train_m1["distance_km"].mean():.1f}km')
print(f'Moyenne train après : {X_train_m1_scaled[:,0].mean():.4f} (≈0)')
print()
print('Scaler sauvegardé pour usage en production')

os.makedirs('../models', exist_ok=True)
joblib.dump(scaler, '../models/scaler.joblib')
joblib.dump(le,     '../models/label_encoder_vehicule.joblib')
print('Sauvegardé : models/scaler.joblib + models/label_encoder_vehicule.joblib')
````

# SECTION 3 — Clustering des corridors

**Objectif :** Regrouper les 43 782 corridors en familles naturelles **sans utiliser le label `is_substitutable`** (apprentissage non supervisé). On compare 3 algorithmes : K-Means, GMM et DBSCAN, puis on valide les clusters obtenus via un dendrogramme et une carte géographique.

## 3.1 Comparaison K-Means vs GMM vs DBSCAN

**Pourquoi comparer 3 algorithmes :**  
Chaque algorithme repose sur des hypothèses différentes sur la forme des clusters. Les tester tous les trois permet de choisir objectivement le plus adapté à nos données.

- **K-Means** : minimise l'inertie intra-cluster. Adapté quand les clusters sont de tailles variables et non-gaussiens.  
- **GMM** (Gaussian Mixture Model) : suppose que chaque cluster suit une distribution gaussienne de taille équilibrée → inadapté ici (38 775 vs 71 corridors).  
- **DBSCAN** : cherche des zones denses séparées par du vide → inadapté ici car nos corridors forment un espace continu.

**Métriques d'évaluation :**  
- **Silhouette Score** : mesure la cohésion interne des clusters (de -1 à 1, plus c'est élevé, mieux c'est)  
- **BIC** (GMM uniquement) : critère bayésien — plus c'est bas, meilleur est le modèle  
- **Elbow Method** (K-Means) : recherche du coude dans la courbe d'inertie

**Résultats :**

| Méthode | k | Silhouette | Décision |
|---|---|---|---|
| **K-Means** | **4** | **0.652** | **Retenu** |
| GMM | 2 | 0.381 | Écarté — négatif dès k=3 |
| DBSCAN | 13 | 0.366 | Écarté — trop fragmenté |

````python
RANDOM_STATE = 42

# Chargement et feature engineering
df_clust = pd.read_csv('../donnee/staging_fact_route_analysis.csv')
df_clust = df_clust.dropna(subset=['co2_train_kg', 'co2_saved_kg'])
df_clust['co2_avion_kg'] = df_clust['co2_avion_kg'].fillna(0)
df_clust['ratio_origin'] = (df_clust['origin_station_traffic'] / df_clust['origin_city_population'].replace(0, np.nan)).fillna(0)
df_clust['ratio_dest']   = (df_clust['dest_station_traffic']   / df_clust['dest_city_population'].replace(0, np.nan)).fillna(0)

FEATURES_CLUST = ['distance_km', 'co2_saved_kg', 'co2_train_kg', 'ratio_origin', 'ratio_dest']
X = df_clust[FEATURES_CLUST].fillna(0)
scaler_clust = StandardScaler()
X_scaled = scaler_clust.fit_transform(X)

print(f"Dataset clustering : {len(X)} corridors | Features : {FEATURES_CLUST}")

# K-Means
k_range = range(2, 7)
inertias, silhouettes = [], []
for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, labels))
best_k = k_range.start + silhouettes.index(max(silhouettes))
km_final = KMeans(n_clusters=best_k, random_state=RANDOM_STATE, n_init=10)
df_clust['cluster_kmeans'] = km_final.fit_predict(X_scaled)

# GMM
bics, gmm_silhouettes = [], []
for k in k_range:
    gmm = GaussianMixture(n_components=k, random_state=RANDOM_STATE, n_init=5)
    gmm_labels = gmm.fit_predict(X_scaled)
    bics.append(gmm.bic(X_scaled))
    gmm_silhouettes.append(silhouette_score(X_scaled, gmm_labels))
best_k_gmm = k_range.start + gmm_silhouettes.index(max(gmm_silhouettes))
gmm_final = GaussianMixture(n_components=best_k_gmm, random_state=RANDOM_STATE, n_init=5)
df_clust['cluster_gmm'] = gmm_final.fit_predict(X_scaled)

# DBSCAN
db = DBSCAN(eps=0.8, min_samples=10)
df_clust['cluster_dbscan'] = db.fit_predict(X_scaled)
n_clusters_db = len(set(df_clust['cluster_dbscan'])) - (1 if -1 in df_clust['cluster_dbscan'].values else 0)
n_noise = (df_clust['cluster_dbscan'] == -1).sum()

# Métriques finales
sil_km  = silhouette_score(X_scaled, df_clust['cluster_kmeans'])
sil_gmm = silhouette_score(X_scaled, df_clust['cluster_gmm'])
mask_db = df_clust['cluster_dbscan'] != -1
sil_db  = silhouette_score(X_scaled[mask_db], df_clust['cluster_dbscan'][mask_db]) if n_clusters_db > 1 else 0

print(f"\n=== Comparaison K-Means vs GMM vs DBSCAN ===")
print(f"  K-Means  k={best_k}   Silhouette={sil_km:.3f}  ← retenu")
print(f"  GMM      k={best_k_gmm}   Silhouette={sil_gmm:.3f}")
print(f"  DBSCAN   k={n_clusters_db}   Silhouette={sil_db:.3f}  ({n_noise} pts bruit)")

# Graphique
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)
colors_km  = cm.Set1(np.linspace(0, 0.8, best_k))
colors_gmm_p = cm.Set2(np.linspace(0, 0.8, best_k_gmm))

fig, axes = plt.subplots(2, 3, figsize=(20, 12))

# Elbow
axes[0,0].plot(list(k_range), inertias, 'bo-', linewidth=2, markersize=8)
axes[0,0].axvline(best_k, color='red', linestyle='--', label=f'k optimal={best_k}')
axes[0,0].set_xlabel('k'); axes[0,0].set_ylabel('Inertie')
axes[0,0].set_title('K-Means — Elbow Method'); axes[0,0].legend()

# Silhouette comparé
x_pos = list(k_range); w = 0.35
axes[0,1].bar([v - w/2 for v in x_pos], silhouettes,     width=w, label='K-Means',
              color=['red' if k == best_k else '#3498db' for k in k_range])
axes[0,1].bar([v + w/2 for v in x_pos], gmm_silhouettes, width=w, label='GMM',
              color=['tomato' if k == best_k_gmm else '#e67e22' for k in k_range])
axes[0,1].set_xlabel('k'); axes[0,1].set_ylabel('Silhouette Score')
axes[0,1].set_title('Silhouette : K-Means vs GMM'); axes[0,1].legend()

# BIC
axes[0,2].plot(list(k_range), bics, 'gs-', linewidth=2, markersize=8)
axes[0,2].axvline(best_k_gmm, color='red', linestyle='--', label=f'k GMM={best_k_gmm}')
axes[0,2].set_xlabel('k'); axes[0,2].set_ylabel('BIC')
axes[0,2].set_title('GMM — BIC Score'); axes[0,2].legend()

# PCA K-Means
profile = df_clust.groupby('cluster_kmeans').agg(
    pct_sub=('is_substitutable','mean'), dist_moy=('distance_km','mean')).round(2)
for c, color in zip(range(best_k), colors_km):
    mask = df_clust['cluster_kmeans'] == c
    lbl  = f"C{c}: {'LH' if profile.loc[c,'pct_sub']<0.5 else 'Sub'} ({mask.sum()})"
    axes[1,0].scatter(X_pca[mask,0], X_pca[mask,1], c=[color], label=lbl, alpha=0.5, s=12)
axes[1,0].set_title(f'K-Means k={best_k} (PCA) — Silhouette={sil_km:.3f}')
axes[1,0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1,0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1,0].legend(fontsize=7)

# PCA GMM
for c, color in zip(range(best_k_gmm), colors_gmm_p):
    mask = df_clust['cluster_gmm'] == c
    axes[1,1].scatter(X_pca[mask,0], X_pca[mask,1], c=[color], label=f"C{c}", alpha=0.5, s=12)
axes[1,1].set_title(f'GMM k={best_k_gmm} (PCA) — Silhouette={sil_gmm:.3f}')
axes[1,1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1,1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1,1].legend(fontsize=8)

# Distance vs CO2
for c, color in zip(range(best_k), colors_km):
    mask = df_clust['cluster_kmeans'] == c
    axes[1,2].scatter(df_clust.loc[mask,'distance_km'], df_clust.loc[mask,'co2_saved_kg'],
                      c=[color], label=f"C{c}", alpha=0.5, s=12)
axes[1,2].axvline(600, color='black', linestyle='--', linewidth=1.5, label='Seuil 600km')
axes[1,2].set_xlabel('Distance (km)'); axes[1,2].set_ylabel('CO2 économisé (kg)')
axes[1,2].set_title('Distance vs CO2 économisé (K-Means)'); axes[1,2].legend(fontsize=8)

plt.suptitle(f'Clustering corridors — K-Means k={best_k} vs GMM k={best_k_gmm} vs DBSCAN k={n_clusters_db}', fontsize=13)
plt.tight_layout()
plt.savefig('../docs/fig_clustering_corridors.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nPCA variance expliquée : {pca.explained_variance_ratio_.sum()*100:.1f}%")
````

## 3.3 Dendrogramme — Validation hiérarchique de k=4

**Pourquoi un dendrogramme :**  
Le dendrogramme est une **validation indépendante** du choix k=4. Il utilise une méthode radicalement différente de K-Means (clustering hiérarchique agglomératif, méthode de Ward) — si les deux méthodes convergent vers le même k, c'est une preuve forte que la structure en 4 groupes est réelle dans les données.

**Méthode Ward :**  
À chaque étape, fusionne les deux clusters qui minimisent l'augmentation de la variance intra-cluster. C'est la méthode hiérarchique qui produit les clusters les plus compacts et homogènes.

**Lecture du dendrogramme :**  
- L'axe Y représente la **distance de fusion** (plus c'est haut, plus les clusters fusionnés sont dissemblables)
- La **ligne rouge pointillée** indique le seuil de coupure → donne 4 clusters
- La position du grand saut de distance confirme le nombre naturel de groupes

**Conclusion :**  
Le dendrogramme confirme k=4 **sans utiliser K-Means**, ce qui valide que la structure observée n'est pas un artefact algorithmique mais reflète une segmentation naturelle des corridors ferroviaires européens.

````python
sample_idx, _ = train_test_split(
    np.arange(len(X_scaled)),
    train_size=500,
    stratify=df_clust['cluster_kmeans'],
    random_state=42
)
X_sample = X_scaled[sample_idx]

# Clustering hiérarchique Ward
Z = linkage(X_sample, method='ward')

# Trouver le seuil de coupure à k=4 (distance maximale avant fusion à 4 clusters → 3)
# last_merges = distances des 20 dernières fusions
last_merges = Z[-20:, 2]
# Seuil = mi-chemin entre la fusion qui donne 5→4 et celle qui donne 4→3
thresh = (Z[-(4-1), 2] + Z[-(4), 2]) / 2

n_sample = len(X_sample)

def label_taille_groupe(leaf_id):
    # leaf_id < n_sample : feuille = un seul point original -> groupe de taille 1
    # leaf_id >= n_sample : feuille contractee = noeud de fusion, taille lue dans Z[:,3]
    if leaf_id < n_sample:
        return '(1)'
    return f'({int(Z[leaf_id - n_sample, 3])})'

fig, ax = plt.subplots(figsize=(14, 6))
dendrogram(
    Z,
    truncate_mode='lastp',
    p=20,
    leaf_rotation=45,
    leaf_font_size=9,
    show_contracted=True,
    leaf_label_func=label_taille_groupe,
    ax=ax
)
ax.axhline(thresh, color='red', linestyle='--', linewidth=2,
           label=f'Seuil k=4 (distance={thresh:.1f})')
ax.set_title('Dendrogramme — Clustering hiérarchique Ward (échantillon n=500)', fontsize=13)
ax.set_xlabel('Corridors (groupes contractés)')
ax.set_ylabel('Distance de fusion (Ward)')
ax.legend(fontsize=10)
plt.tight_layout()
plt.savefig('../docs/fig_dendrogram.png', dpi=150, bbox_inches='tight')
plt.show()

# Validation : nombre de clusters au seuil choisi
labels_hier = fcluster(Z, t=thresh, criterion='distance')
k_hier = len(np.unique(labels_hier))
print(f"Clusters identifiés par coupure Ward : {k_hier}")
print(f"→ Confirme indépendamment k={k_hier} — cohérent avec K-Means (Silhouette={sil_km:.3f})")
print(f"\nFigure sauvegardée : docs/fig_dendrogram.png")
````

````python
# 1) Retrouver, pour les 20 feuilles affichées, lesquelles sont des singletons
R = dendrogram(Z, truncate_mode='lastp', p=20, no_plot=True)
isoles_leaf_ids = [leaf_id for leaf_id in R['leaves'] if leaf_id < n_sample]
print(f"Nombre de points isolés parmi les 20 feuilles affichées : {len(isoles_leaf_ids)}")

# 2) Pour chaque isolé : son corridor d'origine + son cluster K-Means + ses z-scores
lignes_isoles = []
for leaf_id in isoles_leaf_ids:
    idx_original = sample_idx[leaf_id]                  # position dans df_clust
    row = df_clust.iloc[idx_original]
    z_scores = pd.Series(X_scaled[idx_original], index=FEATURES_CLUST)
    feature_extreme = z_scores.abs().idxmax()
    lignes_isoles.append({
        'type': 'ISOLÉ',
        'origin': row['origin'], 'destination': row['destination'],
        'cluster_kmeans': row['cluster_kmeans'],
        **{f'z_{f}': round(z_scores[f], 2) for f in FEATURES_CLUST},
        'feature_la_plus_extreme': feature_extreme,
    })

# 3) Pour comparaison : un trajet "représentatif" par cluster K-Means
lignes_representatifs = []
for k in range(best_k):
    centre = km_final.cluster_centers_[k]
    mask_k = df_clust['cluster_kmeans'] == k
    dists = np.linalg.norm(X_scaled[mask_k.values] - centre, axis=1)
    idx_representatif = df_clust[mask_k].index[dists.argmin()]
    row = df_clust.loc[idx_representatif]
    z_scores = pd.Series(X_scaled[df_clust.index.get_loc(idx_representatif)], index=FEATURES_CLUST)
    lignes_representatifs.append({
        'type': f'Représentatif cluster {k}',
        'origin': row['origin'], 'destination': row['destination'],
        'cluster_kmeans': row['cluster_kmeans'],
        **{f'z_{f}': round(z_scores[f], 2) for f in FEATURES_CLUST},
        'feature_la_plus_extreme': '-',
    })

comparaison = pd.DataFrame(lignes_representatifs + lignes_isoles)
comparaison
````

## 3.4 Interprétation du dendrogramme

**Lecture du résultat obtenu :**
Sur l'échantillon de 500 corridors, la coupure Ward identifie **4 clusters** (`k_hier=4`), avec une Silhouette K-Means de **0.652** sur les 46 106 corridors complets. Les deux méthodes — agglomérative (Ward, 500 points) et partitionnelle (K-Means, 46 106 points) — convergent vers le même nombre de groupes malgré des données et des algorithmes complètement différents : c'est la preuve la plus solide qu'on peut apporter que **k=4 n'est pas un artefact de l'algorithme** mais une structure réelle des corridors ferroviaires.

**Sur les couleurs des branches :**
`dendrogram()` colore automatiquement les branches **sous la ligne rouge** (le seuil de coupure) avec le cycle de couleurs par défaut de matplotlib, une couleur par cluster détecté. Avec k=4, on obtient donc 4 couleurs distinctes — c'est purement un repère visuel pour distinguer les groupes Ward, et son numéro de cluster **ne correspond pas** à la numérotation `cluster_kmeans` / `CLUSTER_COLORS` utilisée sur la carte Folium : ce sont deux clusterings indépendants, exécutés sur des sous-ensembles de points différents.

**Sur `truncate_mode='lastp', p=20` :**
Avec 500 points, l'arbre complet contient 499 fusions — illisible. La troncature n'affiche que les **20 dernières fusions** (le sommet de l'arbre, là où se jouent les regroupements en grands clusters) et contracte tout le reste en feuilles marquées par un point (`show_contracted=True`). On perd donc la granularité fine (les micro-fusions entre points très proches), mais ce n'est pas ce qu'on veut valider ici — seule la structure macro (le nombre de groupes naturels) nous intéresse.

**Limite à garder en tête :**
Le dendrogramme valide la structure sur un **échantillon de 500/46 106 corridors (≈1.1%)**. Ce n'est pas un second clustering complet, juste un test de cohérence à plus petite échelle. Un échantillon différent (autre `seed`) pourrait en théorie donner une coupure légèrement différente — c'est pour ça qu'on parle de *validation indépendante*, pas de preuve formelle, et qu'on s'appuie en priorité sur le K-Means (qui voit, lui, tous les corridors).

**Effet de la stratification par `cluster_kmeans` :**
Après stratification (échantillon respectant les proportions des 4 clusters K-Means), le nombre de feuilles isolées **(1)** est passé de 1 à **3**. La stratification n'a donc pas "dilué" l'isolement, elle l'a révélé plus largement : avec une meilleure représentation de chaque cluster, on voit apparaître plusieurs points qui restent seuls jusqu'au sommet de l'arbre. Ce n'était donc pas un coup de chance du tirage aléatoire initial — ce sont de vrais outliers multivariés, présents dans plusieurs clusters.

**Diagnostic chiffré des 3 isolés (tableau de comparaison) :**
Le tableau croise un trajet *représentatif* par cluster (le point le plus proche de son centroïde K-Means) avec les 3 corridors isolés, et leurs z-scores sur les 5 features du clustering (`X_scaled` est déjà standardisé, donc chaque valeur est directement un z-score) :

| Corridor isolé | Cluster K-Means | Feature la plus extrême | z-score |
|---|---|---|---|
| Tilly → Habay | 3 | `ratio_origin` | **+27.9** |
| Avallon → Gare de Montereau | 2 | `ratio_dest` | **+25.8** |
| Gare de Saint-Florentin-Vergigny → Belleville-sur-Saône | 1 | `ratio_dest` | +3.7 |

Sur les 5 features, **4 restent dans une plage banale (|z| < 0.6)** pour les 3 isolés — seul `ratio_origin` ou `ratio_dest` explose. Ward refuse de les fusionner tôt précisément parce que la distance euclidienne (et donc l'augmentation de variance) est dominée par cette unique dimension extrême : peu importe que les 4 autres features soient ordinaires, un z-score de +28 sur une dimension suffit à isoler le point de tout le reste de l'échantillon. `ratio_origin`/`ratio_dest` = trafic gare / population ville — un grand trafic ferroviaire rapporté à une **très petite ville** fait exploser ce ratio (cas typique d'une gare régionale desservant une commune minuscule).

**Nuance importante : la taille des clusters explique en grande partie ces z-scores extrêmes.**
La répartition réelle des 4 clusters K-Means est très déséquilibrée : **C0 = 4 865**, **C1 = 38 775**, **C2 = 84**, **C3 = 71** corridors. C0 et C1 regroupent à eux seuls **94,7 %** des corridors — ce sont les clusters "normaux" et stables. C2 et C3 ne représentent que **0,33 %** du dataset : il est cohérent, et même attendu, que K-Means leur ait taillé des groupes minuscules et homogènes précisément *parce que* ces corridors partagent un `ratio_origin`/`ratio_dest` atypique (gare à fort trafic desservant une très petite ville). Le fait que le représentant "le plus central" de C2 (`z_ratio_dest=+21.4`) et de C3 (`z_ratio_origin=+23.5`) reste extrême au sens global n'est donc pas la preuve d'un clustering défaillant — c'est la signature attendue d'une **niche réelle** : à l'intérieur de C2/C3, ces points sont probablement homogènes entre eux, même s'ils restent des outliers par rapport à l'ensemble des 46 106 corridors.

**Ce qui reste à trancher :** C2/C3 sont-ils une **niche métier pertinente** (gares régionales sur-desservies, à traiter comme un segment à part dans le M1/M2) ou un **artefact de sensibilité aux outliers** de K-Means/Ward qui se réorganiserait différemment avec une transformation des ratios ? Les deux hypothèses restent ouvertes avec les éléments actuels.

**Piste à tester (pas une correction obligatoire) :**
Appliquer `np.log1p()` ou un *clipping* au 99ᵉ percentile sur `ratio_origin`/`ratio_dest`, puis relancer K-Means, permettrait de vérifier si C2/C3 se redessinent autrement. Si la même niche réapparaît avec des clusters de taille comparable, c'est un signal fort qu'il s'agit d'une vraie structure métier et non d'un artefact — sinon, ça confirmera une sensibilité aux outliers à corriger avant de figer le clustering final.

## 3.5 Comparaison complete : clustering avec ratios transformes (`log1p`)

**Pourquoi ce test :**
On a identifié que `ratio_origin`/`ratio_dest` ont une distribution tres asymetrique (z-scores > +20 sur quelques corridors), et que les clusters C2 (84 corridors) et C3 (71 corridors) semblent largement definis par ces valeurs extrêmes. Reste a savoir si C2/C3 representent une **vraie niche metier** (gares sur-desservies par rapport a la population de leur ville) ou un **artefact de sensibilite aux outliers** de K-Means.

**Protocole :**
On reproduit *exactement* les mêmes pipelines (Elbow, Silhouette K-Means vs GMM, BIC, DBSCAN, projection PCA et dendrogramme Ward stratifie + diagnostic des points isolés), mais en appliquant `np.log1p()` sur `ratio_origin`/`ratio_dest` avant la standardisation. Toutes les variables sont suffixées `_log` pour ne pas écraser les résultats originaux : la comparaison se fait côte à côte, pas après remplacement.

- Si une niche de taille comparable a C2/C3 réapparait avec une silhouette équivalente -> **vraie structure métier**, le clustering d'origine était fiable.
- Si les clusters se redistribuent très différemment (niche qui disparait, fusionne avec C0/C1, silhouette qui chute) -> **artefact d'outliers**, à corriger avant de figer le clustering final.

````python
RANDOM_STATE = 42

# Meme feature engineering que df_clust, mais ratio_origin/ratio_dest passes en log1p
# (log1p compresse les grandes valeurs sans affecter les petites : log1p(0)=0)
df_clust_log = df_clust.copy()
df_clust_log['ratio_origin'] = np.log1p(df_clust_log['ratio_origin'])
df_clust_log['ratio_dest']   = np.log1p(df_clust_log['ratio_dest'])

X_log = df_clust_log[FEATURES_CLUST].fillna(0)
scaler_clust_log = StandardScaler()
X_scaled_log = scaler_clust_log.fit_transform(X_log)

print(f"Dataset clustering (log1p) : {len(X_log)} corridors | Features : {FEATURES_CLUST}")

# K-Means
k_range = range(2, 7)
inertias_log, silhouettes_log = [], []
for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_scaled_log)
    inertias_log.append(km.inertia_)
    silhouettes_log.append(silhouette_score(X_scaled_log, labels))
best_k_log = k_range.start + silhouettes_log.index(max(silhouettes_log))
km_final_log = KMeans(n_clusters=best_k_log, random_state=RANDOM_STATE, n_init=10)
df_clust_log['cluster_kmeans_log'] = km_final_log.fit_predict(X_scaled_log)

# GMM
bics_log, gmm_silhouettes_log = [], []
for k in k_range:
    gmm = GaussianMixture(n_components=k, random_state=RANDOM_STATE, n_init=5)
    gmm_labels = gmm.fit_predict(X_scaled_log)
    bics_log.append(gmm.bic(X_scaled_log))
    gmm_silhouettes_log.append(silhouette_score(X_scaled_log, gmm_labels))
best_k_gmm_log = k_range.start + gmm_silhouettes_log.index(max(gmm_silhouettes_log))
gmm_final_log = GaussianMixture(n_components=best_k_gmm_log, random_state=RANDOM_STATE, n_init=5)
df_clust_log['cluster_gmm_log'] = gmm_final_log.fit_predict(X_scaled_log)

# DBSCAN
db_log = DBSCAN(eps=0.8, min_samples=10)
df_clust_log['cluster_dbscan_log'] = db_log.fit_predict(X_scaled_log)
n_clusters_db_log = len(set(df_clust_log['cluster_dbscan_log'])) - (1 if -1 in df_clust_log['cluster_dbscan_log'].values else 0)
n_noise_log = (df_clust_log['cluster_dbscan_log'] == -1).sum()

# Metriques finales
sil_km_log  = silhouette_score(X_scaled_log, df_clust_log['cluster_kmeans_log'])
sil_gmm_log = silhouette_score(X_scaled_log, df_clust_log['cluster_gmm_log'])
mask_db_log = df_clust_log['cluster_dbscan_log'] != -1
sil_db_log  = silhouette_score(X_scaled_log[mask_db_log], df_clust_log['cluster_dbscan_log'][mask_db_log]) if n_clusters_db_log > 1 else 0

print(f"\n=== Comparaison K-Means vs GMM vs DBSCAN (ratios en log1p) ===")
print(f"  K-Means  k={best_k_log}   Silhouette={sil_km_log:.3f}  <- retenu")
print(f"  GMM      k={best_k_gmm_log}   Silhouette={sil_gmm_log:.3f}")
print(f"  DBSCAN   k={n_clusters_db_log}   Silhouette={sil_db_log:.3f}  ({n_noise_log} pts bruit)")

# Graphique (meme mise en page qu'en 3.1, suffixe _log)
pca_log = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca_log = pca_log.fit_transform(X_scaled_log)
colors_km_log  = cm.Set1(np.linspace(0, 0.8, best_k_log))
colors_gmm_log = cm.Set2(np.linspace(0, 0.8, best_k_gmm_log))

fig, axes = plt.subplots(2, 3, figsize=(20, 12))

axes[0,0].plot(list(k_range), inertias_log, 'bo-', linewidth=2, markersize=8)
axes[0,0].axvline(best_k_log, color='red', linestyle='--', label=f'k optimal={best_k_log}')
axes[0,0].set_xlabel('k'); axes[0,0].set_ylabel('Inertie')
axes[0,0].set_title('K-Means (log1p) - Elbow Method'); axes[0,0].legend()

x_pos = list(k_range); w = 0.35
axes[0,1].bar([v - w/2 for v in x_pos], silhouettes_log,     width=w, label='K-Means',
              color=['red' if k == best_k_log else '#3498db' for k in k_range])
axes[0,1].bar([v + w/2 for v in x_pos], gmm_silhouettes_log, width=w, label='GMM',
              color=['tomato' if k == best_k_gmm_log else '#e67e22' for k in k_range])
axes[0,1].set_xlabel('k'); axes[0,1].set_ylabel('Silhouette Score')
axes[0,1].set_title('Silhouette (log1p) : K-Means vs GMM'); axes[0,1].legend()

axes[0,2].plot(list(k_range), bics_log, 'gs-', linewidth=2, markersize=8)
axes[0,2].axvline(best_k_gmm_log, color='red', linestyle='--', label=f'k GMM={best_k_gmm_log}')
axes[0,2].set_xlabel('k'); axes[0,2].set_ylabel('BIC')
axes[0,2].set_title('GMM (log1p) - BIC Score'); axes[0,2].legend()

for c, color in zip(range(best_k_log), colors_km_log):
    mask = df_clust_log['cluster_kmeans_log'] == c
    axes[1,0].scatter(X_pca_log[mask,0], X_pca_log[mask,1], c=[color], label=f"C{c} ({mask.sum()})", alpha=0.5, s=12)
axes[1,0].set_title(f'K-Means (log1p) k={best_k_log} (PCA) - Silhouette={sil_km_log:.3f}')
axes[1,0].set_xlabel(f'PC1 ({pca_log.explained_variance_ratio_[0]*100:.1f}%)')
axes[1,0].set_ylabel(f'PC2 ({pca_log.explained_variance_ratio_[1]*100:.1f}%)')
axes[1,0].legend(fontsize=7)

for c, color in zip(range(best_k_gmm_log), colors_gmm_log):
    mask = df_clust_log['cluster_gmm_log'] == c
    axes[1,1].scatter(X_pca_log[mask,0], X_pca_log[mask,1], c=[color], label=f"C{c}", alpha=0.5, s=12)
axes[1,1].set_title(f'GMM (log1p) k={best_k_gmm_log} (PCA) - Silhouette={sil_gmm_log:.3f}')
axes[1,1].set_xlabel(f'PC1 ({pca_log.explained_variance_ratio_[0]*100:.1f}%)')
axes[1,1].set_ylabel(f'PC2 ({pca_log.explained_variance_ratio_[1]*100:.1f}%)')
axes[1,1].legend(fontsize=8)

for c, color in zip(range(best_k_log), colors_km_log):
    mask = df_clust_log['cluster_kmeans_log'] == c
    axes[1,2].scatter(df_clust_log.loc[mask,'distance_km'], df_clust_log.loc[mask,'co2_saved_kg'],
                      c=[color], label=f"C{c}", alpha=0.5, s=12)
axes[1,2].axvline(600, color='black', linestyle='--', linewidth=1.5, label='Seuil 600km')
axes[1,2].set_xlabel('Distance (km)'); axes[1,2].set_ylabel('CO2 economise (kg)')
axes[1,2].set_title('Distance vs CO2 economise (K-Means log1p)'); axes[1,2].legend(fontsize=8)

plt.suptitle(f'Clustering corridors (ratios log1p) - K-Means k={best_k_log} vs GMM k={best_k_gmm_log} vs DBSCAN k={n_clusters_db_log}', fontsize=13)
plt.tight_layout()
plt.savefig('../docs/fig_clustering_corridors_log1p.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nPCA variance expliquee (log1p) : {pca_log.explained_variance_ratio_.sum()*100:.1f}%")

# Comparaison directe des tailles de clusters : original vs log1p
tailles_originales = df_clust['cluster_kmeans'].value_counts().sort_index()
tailles_log = df_clust_log['cluster_kmeans_log'].value_counts().sort_index()
print("\n=== Comparaison des tailles de clusters : ratios bruts vs log1p ===")
print(pd.DataFrame({'Original (ratios bruts)': tailles_originales, 'Apres log1p': tailles_log}))
````

## 3.6 Dendrogramme (log1p) - validation hierarchique sur les ratios transformes

Meme principe : echantillon stratifie de 500 corridors (sur `cluster_kmeans_log` cette fois), clustering hiérarchique Ward, et diagnostic des points isolés, mais sur `X_scaled_log` (ratios en `log1p`). Si la transformation a bien résolu le problème d'échelle, on doit voir **moins de points isolés** (ou des z-scores extrêmes nettement réduits).

````python
sample_idx_log, _ = train_test_split(
    np.arange(len(X_scaled_log)),
    train_size=500,
    stratify=df_clust_log['cluster_kmeans_log'],
    random_state=42
)
X_sample_log = X_scaled_log[sample_idx_log]

Z_log = linkage(X_sample_log, method='ward')

n_sample_log = len(X_sample_log)
last_merges_log = Z_log[-20:, 2]
thresh_log = (Z_log[-(best_k_log-1), 2] + Z_log[-(best_k_log), 2]) / 2

def label_taille_groupe_log(leaf_id):
    if leaf_id < n_sample_log:
        return '(1)'
    return f'({int(Z_log[leaf_id - n_sample_log, 3])})'

fig, ax = plt.subplots(figsize=(14, 6))
dendrogram(
    Z_log,
    truncate_mode='lastp',
    p=20,
    leaf_rotation=45,
    leaf_font_size=9,
    show_contracted=True,
    leaf_label_func=label_taille_groupe_log,
    ax=ax
)
ax.axhline(thresh_log, color='red', linestyle='--', linewidth=2,
           label=f'Seuil k={best_k_log} (distance={thresh_log:.1f})')
ax.set_title('Dendrogramme (log1p) - Clustering hierarchique Ward (echantillon n=500)', fontsize=13)
ax.set_xlabel('Corridors (groupes contractes)')
ax.set_ylabel('Distance de fusion (Ward)')
ax.legend(fontsize=10)
plt.tight_layout()
plt.savefig('../docs/fig_dendrogram_log1p.png', dpi=150, bbox_inches='tight')
plt.show()

labels_hier_log = fcluster(Z_log, t=thresh_log, criterion='distance')
k_hier_log = len(np.unique(labels_hier_log))
print(f"Clusters identifies par coupure Ward (log1p) : {k_hier_log}")
print(f"-> Comparer a k_hier={k_hier} (ratios bruts, paragraphe 3.3)")

# Diagnostic des points isoles, identique au 3.4 mais sur la version log1p
R_log = dendrogram(Z_log, truncate_mode='lastp', p=20, no_plot=True)
isoles_leaf_ids_log = [leaf_id for leaf_id in R_log['leaves'] if leaf_id < n_sample_log]
print(f"\nNombre de points isoles parmi les 20 feuilles affichees (log1p) : {len(isoles_leaf_ids_log)}")
print("-> Comparer a 3 isoles trouves avec les ratios bruts (paragraphe 3.4)")

lignes_isoles_log = []
for leaf_id in isoles_leaf_ids_log:
    idx_original = sample_idx_log[leaf_id]
    row = df_clust_log.iloc[idx_original]
    z_scores = pd.Series(X_scaled_log[idx_original], index=FEATURES_CLUST)
    feature_extreme = z_scores.abs().idxmax()
    lignes_isoles_log.append({
        'origin': row['origin'], 'destination': row['destination'],
        'cluster_kmeans_log': row['cluster_kmeans_log'],
        **{f'z_{f}': round(z_scores[f], 2) for f in FEATURES_CLUST},
        'feature_la_plus_extreme': feature_extreme,
    })

if lignes_isoles_log:
    display(pd.DataFrame(lignes_isoles_log))
else:
    print("Aucun point isole apres transformation log1p.")
````

## 3.7 Conclusion : `log1p` retenu comme clustering de référence

**Verdict sur C2/C3 :** les trois preuves convergentes (disparition des clusters sans réapparition équivalente, passage de 3 à 0 points isolés dans le dendrogramme, convergence K-Means/Ward sur un nouveau k) confirment que C2 (84 corridors) et C3 (71 corridors) étaient un **artefact de sensibilité aux outliers** de `ratio_origin`/`ratio_dest`, et non une niche métier réelle. Avec une distribution aussi asymétrique (z-scores bruts > +20), l'écart-type global était gonflé par une poignée de gares à très faible population de ville, ce qui dominait la distance euclidienne utilisée par K-Means et par Ward.

**Comparatif des deux versions :**

| | Ratios bruts | Ratios `log1p` |
|---|---|---|
| k retenu (K-Means = Ward) | 4 | **2** |
| Silhouette K-Means | 0.652 | 0.499 |
| Points isolés (dendrogramme) | 3 | **0** |
| Tailles de clusters | 4 852 / 38 775 / 84 / 71 | **5 320 / 38 462** |

La baisse de silhouette (0.652 → 0.499) n'est pas une régression : le score original était artificiellement gonflé par la séparation triviale de C2/C3, des clusters minuscules et extrêmement éloignés du reste. Une fois ces outliers neutralisés, 0.499 reflète honnêtement la structure réelle des données (et reste une silhouette correcte, au-dessus du seuil usuel de 0.4 pour des données réelles bruitées).

**Cohérence avec le problème métier :** k=2 colle plus naturellement à la variable binaire `is_substitutable` du Modèle 1 que les 4 sous-groupes précédents, qui sur-segmentaient surtout à cause du bruit des ratios. La nouvelle frontière à 2 clusters (5 320 / 38 462) est très proche de l'ancienne opposition "Long-haul peu substituable" (ancien C0=4 852) vs "Standard substituable" (ancien C1+C2+C3=38 930) : le clustering `log1p` ne change pas la lecture métier, il la **simplifie et la fiabilise** en absorbant les 155 corridors (0,33%) qui faussaient artificiellement la segmentation en 4 groupes.

**Recommandation :** retenir la version `log1p` (`df_clust_log`, `cluster_kmeans_log`, k=2) comme clustering de référence pour la suite du projet (M3 / segmentation), et conserver la version à 4 clusters uniquement comme illustration pédagogique de l'effet des outliers sur un clustering non robuste.

## 3.8 Coherence avec les donnees d'entree : le clustering recoupe-t-il `is_substitutable` ?

**Question posee :** la proportion globale `is_substitutable` (~84% substituables / ~16% non) est-elle proche de la proportion des 2 clusters `log1p` (5 320 / 38 462) ? Et si oui, est-ce que le clustering redécouvre simplement le label ?

````python
# Croisement cluster_kmeans_log x is_substitutable : le clustering recoupe-t-il le label ?
prop_label = df_clust_log['is_substitutable'].value_counts(normalize=True).round(4) * 100
prop_cluster = df_clust_log['cluster_kmeans_log'].value_counts(normalize=True).round(4) * 100

print("Proportion globale is_substitutable :")
print(prop_label)
print()
print("Proportion des 2 clusters (log1p) :")
print(prop_cluster)
print()

crosstab_pct = pd.crosstab(df_clust_log['cluster_kmeans_log'], df_clust_log['is_substitutable'],
                            normalize='index').round(4) * 100
crosstab_n = pd.crosstab(df_clust_log['cluster_kmeans_log'], df_clust_log['is_substitutable'])
print("Croisement cluster x is_substitutable (effectifs) :")
print(crosstab_n)
print()
print("Croisement cluster x is_substitutable (% par cluster) :")
print(crosstab_pct)
````

**Résultat du croisement :**

| Cluster | Non-substituable | Substituable | Taux substituable |
|---|---|---|---|
| Cluster 0 (5 320) | 2 364 | 2 956 | **55.6%** |
| Cluster 1 (38 462) | 4 530 | 33 932 | **88.2%** |

**Verdict : ressemblance de proportions, pas redécouverte du label.**
La proportion globale `is_substitutable` (84.25% / 15.75%) et la proportion des 2 clusters (88.0% / 12.2%) sont du même ordre de grandeur, mais pas identiques (écart de ~4 points). Le vrai test est le croisement : si le clustering redécouvrait simplement le label, on attendrait des clusters proches de 0%/100%. Ce n'est pas le cas — le Cluster 0 est presque un coin-flip (55.6% / 44.4%). La ressemblance de proportions globales est donc en grande partie une coïncidence d'échelle, pas une équivalence structurelle.

**Ce que ça révèle de plus intéressant : le clustering retrouve indirectement le profil "Long-haul".**
Le clustering `log1p` n'a vu ni `is_substitutable`, ni les features qui le déterminent le plus (`service_share` 11.6%, `trip_count_corridor` 47.9% d'importance dans le XGBoost du Modèle 1, cf. §1.4b) — seulement `distance_km`, `co2_saved_kg`, `co2_train_kg`, `ratio_origin`, `ratio_dest`. Pourtant :
- **Cluster 0** (5 320, profil proche de l'ancien C0 "Long haul" à 4 852 corridors, §3.2) a un taux de substitution de **55.6%, nettement sous la moyenne globale** (84.25%) — cohérent avec la corrélation négative `distance_km`/`co2_*` ↔ `is_substitutable` déjà observée en §1.5-1.6 (r=-0.31 pour `distance_km`).
- **Cluster 1** (38 462, le reste) a un taux de **88.2%, au-dessus de la moyenne** — cohérent avec les corridors courts/moyens, mieux desservis en moyenne.

**Conclusion :** le clustering géographique/CO2 capte un **signal partiel et indirect** de la substituabilité, sans jamais y avoir accès — ce qui valide une nouvelle fois (de façon totalement indépendante, par une méthode non supervisée) que `distance_km`/`co2_saved_kg` portent une vraie information métier. Mais le mélange 55.6%/44.4% du Cluster 0 confirme aussi la limite : la géographie seule ne suffit pas à trancher la substituabilité — c'est la **desserte ferroviaire** (`service_share`, volontairement exclue de ce clustering) qui reste le facteur réellement déterminant, comme établi dès le début et confirmé par les importances du Modèle 1.
